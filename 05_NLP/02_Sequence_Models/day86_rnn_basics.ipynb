#Introduction to RNNs (Recurrent Neural Networks)

Standard neural networks (like the ones used in Phase 4) assume all inputs are independent. However, language is a sequence—the meaning of a word depends on the words before it. RNNs solve this by using a "hidden state" (memory) that passes information from one time step to the next.

1. Initialize the Environment

RNNs are typically implemented using TensorFlow/Keras. We will start by setting up a basic environment to handle sequential data.

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense, Embedding

# Verify GPU availability for training
print("Num GPUs Available: ", len(tf.config.list_physical_devices('GPU')))
```

2. Build a Basic SimpleRNN Architecture

We will build a model that takes a sequence of word indices and processes them through a Recurrent layer.

```python
# Model Hyperparameters
vocab_size = 1000      # Size of our vocabulary
embedding_dim = 64     # Dimension of the word vectors
sequence_length = 10   # Number of words in each input sequence

model = Sequential([
    # 1. Embedding Layer: Maps word indices to dense vectors
    Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=sequence_length),
    
    # 2. SimpleRNN Layer: The memory unit
    # units=32: The size of the hidden state vector
    SimpleRNN(units=32, activation='tanh'),
    
    # 3. Output Layer: Binary classification (e.g., Sentiment analysis)
    Dense(1, activation='sigmoid')
])

model.summary()
```

3. Understand the Vanishing Gradient Problem

While SimpleRNNs have memory, it is very "short-term." As the input sequence gets longer, the gradients used for training become extremely small (vanish), causing the network to forget the beginning of the sentence.

- Key Insight: This is why SimpleRNNs struggle with long paragraphs and are usually replaced by LSTMs or GRUs in production.
