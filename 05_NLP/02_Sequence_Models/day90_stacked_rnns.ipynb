#Deep RNN Stacking & Sub-Module Wrap-up

To wrap up Sequence Models, we are moving from single layers to Stacked Architectures. Just like a deep CNN has multiple layers to "see" more complex features, a Stacked LSTM uses multiple layers to understand deeper linguistic hierarchies.

**The Golden Rule:** return_sequences=True
When you stack layers, the previous layer must pass the entire sequence (not just the final hidden state) to the next one.


1. Implementation

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Embedding

model = Sequential([
    Embedding(input_dim=1000, output_dim=64, input_length=20),
    
    # Layer 1: MUST return sequences so Layer 2 has input for every word
    LSTM(64, return_sequences=True), 
    
    # Layer 2: Final layer in the stack, so return_sequences=False
    # (unless you were adding a Layer 3!)
    LSTM(32, return_sequences=False),
    
    Dense(1, activation='sigmoid')
])

model.summary()
```
