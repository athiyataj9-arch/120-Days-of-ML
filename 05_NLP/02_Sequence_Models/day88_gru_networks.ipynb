#GRUs (Gated Recurrent Units) — Efficient Sequences

While LSTMs (Day 87) are powerful, they are computationally expensive because they have three gates and two internal states (cell and hidden). GRUs are a streamlined version that merges the cell and hidden states and uses only two gates: Reset and Update. They are faster to train and often perform just as well as LSTMs on smaller datasets.

1. Understand the Two Gates

GRUs simplify the sequence modeling process:

- **Update Gate:** Determines how much of the past information needs to be passed along to the future. It acts similarly to the combination of the Forget and Input gates in an LSTM.

- **Reset Gate:** Decides how much of the past information to forget.

2. Build a GRU Model with Keras

We will implement a GRU architecture. Notice that the number of parameters is lower than the LSTM version, making it more efficient.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense, Embedding, Bidirectional

# Model Hyperparameters
vocab_size = 5000
embedding_dim = 128
max_length = 50

model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=max_length),
    
    # Using GRU instead of LSTM
    # GRUs generally train faster and use less memory
    GRU(units=64, return_sequences=False),
    
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.summary()
```
