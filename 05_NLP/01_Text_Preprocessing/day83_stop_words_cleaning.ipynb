#Stop Words & Basic Text Cleaning

In NLP, Stop Words are common words like "is," "the," "a," and "in." While essential for human grammar, they carry very little unique information for Machine Learning models. Removing them reduces "noise" and allows the model to focus on high-value keywords (e.g., "Machine," "Failure," "Fake").

1. Install & Load Stopwords Corpus

NLTK provides a pre-defined list of stopwords for multiple languages. We must download this collection before filtering our text.

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import string

# Download the stopwords dataset
nltk.download('stopwords')
nltk.download('punkt')

# Load English stopwords
stop_words = set(stopwords.words('english'))

print(f"Total English Stopwords: {len(stop_words)}")
print(f"Sample Stopwords: {list(stop_words)[:10]}")
```

2. Implement Text Cleaning Pipeline

In this step, we combine Tokenization, Lowercasing, and Stopword Removal. We also remove punctuation, which is usually unnecessary for basic sentiment or classification tasks.

```python
def clean_text(raw_text):
    # 1. Lowercase everything
    text = raw_text.lower()
    
    # 2. Tokenize
    tokens = word_tokenize(text)
    
    # 3. Filter out Stopwords AND Punctuation
    cleaned_tokens = [
        word for word in tokens 
        if word not in stop_words and word not in string.punctuation
    ]
    
    return cleaned_tokens

sample_text = "The quick brown fox is jumping over the lazy dog!"
cleaned = clean_text(sample_text)

print(f"Original: {sample_text}")
print(f"Cleaned:  {cleaned}")
```

3. Handling Custom Stopwords

Sometimes, domain-specific words (like "video" for your YouTube project) appear too often and act as noise. We can extend the default list.

```python
custom_stops = stop_words.copy()
custom_stops.add("athiya") # Adding a specific name as a stopword

text_with_name = "Hello, this is Athiya and I am learning NLP."
tokens = word_tokenize(text_with_name.lower())

# Filter using custom list
filtered = [w for w in tokens if w not in custom_stops and w not in string.punctuation]
print(f"Filtered with Custom List: {filtered}")
```

4. The Impact Analysis

Compare the "Information Density" before and after cleaning.

```python
raw_sentence = "This is a very long sentence that contains many words which do not help the model."
cleaned_sentence = clean_text(raw_sentence)

print(f"Raw Length: {len(word_tokenize(raw_sentence))}")
print(f"Cleaned Length: {len(cleaned_sentence)}")
print(f"Reduction: {((len(word_tokenize(raw_sentence)) - len(cleaned_sentence)) / len(word_tokenize(raw_sentence))) * 100:.2f}%")
```
