#Stemming & Lemmatization — Word Normalization

Normalization is the process of converting a word to its base form. This ensures that a model treats "running," "runs," and "ran" as the same concept, significantly reducing the complexity of the vocabulary.

1. Install & Initialize Dependencies

To perform these tasks, we need the WordNet lexical database and the Open Multilingual Wordnet (OMW) to help the Lemmatizer understand dictionary roots.

```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import wordnet

# Download the required dictionaries for normalization
nltk.download('wordnet')
nltk.download('omw-1.4')

# Initialize the tools
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

2. Implement Stemming (The Fast Approach)

Stemming uses a heuristic process to chop off the ends of words. It is fast but often results in "stems" that aren't actual words (e.g., "studies" becomes "studi").

```python
words = ["working", "worked", "works", "studies", "studying"]

stemmed_words = [stemmer.stem(w) for w in words]
print(f"Original: {words}")
print(f"Stemmed:  {stemmed_words}")
```

3. Implement Lemmatization (The Accurate Approach)

Lemmatization uses a vocabulary and morphological analysis to return the "Lemma" (dictionary form). Unlike stemming, "studies" correctly becomes "study," and "feet" becomes "foot."

```python
complex_words = ["feet", "better", "studies", "rocks"]

# Lemmatizing as Nouns (Default)
lemmatized_words = [lemmatizer.lemmatize(w) for w in complex_words]

# Note: For words like 'better', we specify the POS (Part of Speech) as Adjective
better_lemma = lemmatizer.lemmatize("better", pos="a") 

print(f"Complex Words: {complex_words}")
print(f"Lemmatized:    {lemmatized_words}")
print(f"Lemma for 'better' (Adj): {better_lemma}")
```

4. Comparison Analysis 

In this step, we run a side-by-side comparison to see when to use each method.

```python
test_set = ["trouble", "troubling", "troubled", "am", "is", "are"]

print(f"{'Word':<12} | {'Stem':<12} | {'Lemma':<12}")
print("-" * 40)
for w in test_set:
    print(f"{w:<12} | {stemmer.stem(w):<12} | {lemmatizer.lemmatize(w, pos='v'):<12}")
```
