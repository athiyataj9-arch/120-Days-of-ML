#TF-IDF Vectorization — Measuring Word Importance

Computers cannot understand text; they only understand numbers. TF-IDF (Term Frequency-Inverse Document Frequency) is a statistical formula used to convert text into a meaningful vector of numbers. Unlike simple word counting, TF-IDF penalizes common words (like "the") and rewards rare, meaningful words (like "XGBoost" or "RAG").

#1. Understand the Formula 

TF-IDF is calculated by multiplying two metrics:

- Term Frequency (TF): How frequently a word appears in a specific document.
$$TF(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d}$$

- Inverse Document Frequency (IDF): How unique a word is across all documents in your dataset.
$$IDF(t, D) = \log\left(\frac{\text{Total number of documents } D}{\text{Number of documents with term } t \text{ in them}}\right)$$

2. Initialize Scikit-Learn TfidfVectorizer

We use sklearn because it is highly optimized for large datasets and handles the math automatically.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# Sample Corpus (e.g., descriptions of your ML projects)
corpus = [
    "Machine failure prediction using XGBoost and sensor data.",
    "Online fake news detection using NLP and TF-IDF vectorization.",
    "YouTube QA system using RAG and multilingual transcripts.",
    "Predictive maintenance for industrial equipment using Machine Learning."
]

# Initialize the Vectorizer
# stop_words='english' automatically handles the cleaning we did in Day 83
tfidf = TfidfVectorizer(stop_words='english')
```

3. Fit and Transform the Data

In this step, the model "learns" the vocabulary and calculates the importance scores for every word in your corpus.

```python
# Transform the corpus into a TF-IDF matrix
tfidf_matrix = tfidf.fit_transform(corpus)

# Get the feature names (the words in the vocabulary)
feature_names = tfidf.get_feature_names_out()

# Convert to a DataFrame for better visualization
df_tfidf = pd.DataFrame(
    tfidf_matrix.toarray(), 
    columns=feature_names, 
    index=[f"Doc {i+1}" for i in range(len(corpus))]
)

print("TF-IDF Scores Matrix:")
print(df_tfidf.head())
```

4. Analyze Key Terms 

By looking at the scores, you can see which words define each document. A higher score means the word is more "characteristic" of that specific text.

```python
# Find the top 3 words for Document 2 (Fake News Project)
doc_idx = 1 # Index for Document 2
row = df_tfidf.iloc[doc_idx]
print(f"\nTop keywords for Document 2:\n{row.sort_values(ascending=False).head(3)}")
```
