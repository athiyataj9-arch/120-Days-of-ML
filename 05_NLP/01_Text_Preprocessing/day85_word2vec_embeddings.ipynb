#Word Embeddings (Word2Vec) — Contextual Meaning

While TF-IDF (Day 84) measures word importance, it fails to understand similarity. To a TF-IDF model, "King" and "Queen" are as different as "King" and "Apple." Word Embeddings solve this by representing words as dense vectors in a multi-dimensional space where words with similar meanings are physically close to each other.

1. nstall & Load Gensim 

Gensim is the industry-standard library for topic modeling and document similarity. We will use it to implement Word2Vec, a model developed by Google.

```python
import sys
import subprocess

# Ensure Gensim is installed
try:
    import gensim
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "gensim"])
    import gensim

from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
import nltk

nltk.download('punkt')
```

2. Prepare Training Data

Word2Vec learns from the context of words. It requires a list of tokenized sentences. We will simulate a small corpus related to your Data Science background.

```python
# Sample sentences (Corpus)
data = [
    "Machine learning is a subset of artificial intelligence",
    "Deep learning uses neural networks to process data",
    "Natural language processing helps computers understand text",
    "Artificial intelligence and machine learning are transforming industry",
    "Data science involves statistics and probability"
]

# Tokenize the corpus
tokenized_data = [word_tokenize(sentence.lower()) for sentence in data]
```

3. Train the Word2Vec Model 

We define the vector size (dimensions) and the "window" (how many neighboring words to look at).

```python
# vector_size=100: Each word is represented by 100 numbers
# window=5: Look at 5 words before and after the target word
# min_count=1: Ignore words that appear fewer than 1 time
model = Word2Vec(sentences=tokenized_data, vector_size=100, window=5, min_count=1, workers=4)

# Get the vector for a specific word
vector = model.wv['learning']
print(f"Vector for 'learning' (first 10 dims): {vector[:10]}")
```

4. Find Semantic Similarities

The true power of embeddings is finding related concepts. Even with this tiny dataset, the model attempts to find associations.

```python
# Find words most similar to 'machine'
similar_words = model.wv.most_similar('machine', topn=3)

print("Words most similar to 'machine':")
for word, score in similar_words:
    print(f"{word}: {score:.4f}")

# Word Arithmetic (The famous: King - Man + Woman = Queen logic)
# Note: Requires a much larger pre-trained model like Google News vectors
```
