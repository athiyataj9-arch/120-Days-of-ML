#Tokenization (Breaking the Code)

In NLP, Tokenization is the process of breaking a raw string of text into smaller units (tokens) like words or sentences. This is the first step before any machine learning can happen.

1. Environment & Dependency Check

Run this first to ensure nltk is properly linked to your kernel and all models are downloaded.

2. Implementing the Tokenization Suite

We will compare three different ways to "break" text. This is crucial for your YouTube QA System because user comments use different grammar than video transcripts.

```python
from nltk.tokenize import word_tokenize, sent_tokenize, TweetTokenizer

# Sample data: A mix of formal and social media text
text = "Hello! I'm Athiya. Welcome to Phase 5: NLP. Are you ready? 🚀 #120DaysOfML"

# 1. Sentence Tokenization (The 'Macro' View)
# Best for: Summarization and paragraph splitting
sentences = sent_tokenize(text)
print(f"--- Sentences ({len(sentences)}) ---")
print(sentences)

# 2. Standard Word Tokenization (The 'Standard' View)
# Best for: Formal documents and transcripts
words = word_tokenize(text)
print(f"\n--- Standard Words ---")
print(words)

# 3. Tweet Tokenization (The 'Social' View)
# Best for: Handling Emojis, Hashtags, and Slang
t_tokenizer = TweetTokenizer()
tweet_tokens = t_tokenizer.tokenize(text)
print(f"\n--- Tweet/Social Tokens ---")
print(tweet_tokens)
```

#Why this matters for your Projects:

- For your YouTube QA System: Video transcripts often lack punctuation. sent_tokenize helps you recreate logical chunks before feeding them into your RAG pipeline.

- For Fake News Detection: People who spread misinformation often use excessive emojis and hashtags. Using TweetTokenizer ensures your model "sees" the emoji as a feature rather than noise.
