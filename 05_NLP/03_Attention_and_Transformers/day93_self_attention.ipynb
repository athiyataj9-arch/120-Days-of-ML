#Self-Attention — The Core of Transformers

Today is a turning point. We are moving away from RNNs and LSTMs entirely. In Self-Attention, the model doesn't just look at an "Encoder" from a "Decoder." Instead, every word in a single sentence looks at every other word in that same sentence to understand its context.

#Key Concept: Queries, Keys, and Values ($Q, K, V$)

Think of Self-Attention like a search engine:
1. **Query ($Q$):** What I am looking for (the current word).
2. **Key ($K$):** The labels of all words in the sentence (to match against).
3. **Value ($V$):** The actual information contained in those words.
The mathematical formula for Scaled Dot-Product Attention is:
$$\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V$$

#Implementing Scaled Dot-Product Attention

Here is how you can implement this core math in Python:

```python
import numpy as np
import tensorflow as tf

def scaled_dot_product_attention(q, k, v):
    # 1. Dot product of Query and Key
    matmul_qk = tf.matmul(q, k, transpose_b=True)
    
    # 2. Scale by square root of depth
    dk = tf.cast(tf.shape(k)[-1], tf.float32)
    scaled_attention_logits = matmul_qk / tf.math.sqrt(dk)
    
    # 3. Apply Softmax to get weights (0 to 1)
    attention_weights = tf.nn.softmax(scaled_attention_logits, axis=-1)
    
    # 4. Multiply weights by Values
    output = tf.matmul(attention_weights, v)
    
    return output, attention_weights

# Example usage:
temp_q = tf.constant([[10, 0, 0]], dtype=tf.float32)  # Search for 1st word
temp_k = tf.constant([[10, 0, 0], [0, 10, 0], [0, 0, 10]], dtype=tf.float32)
temp_v = tf.constant([[1, 0], [10, 0], [100, 5]], dtype=tf.float32)

out, weights = scaled_dot_product_attention(temp_q, temp_k, temp_v)
print("Attention Weights:", weights.numpy())
print("Output:", out.numpy())
```
