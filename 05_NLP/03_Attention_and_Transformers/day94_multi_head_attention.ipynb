#Multi-Head Attention — Parallelizing Context

Yesterday, we mastered Self-Attention, where a word looks at others through one lens. But what if a word needs to understand multiple relationships at once? For example, in the sentence "The cat sat on the mat", the word "sat" needs to relate to "cat" (the actor) and "mat" (the location) simultaneously.

**Multi-Head Attention** solves this by running multiple "heads" of self-attention in parallel, each focusing on different parts of the sentence.

1. **The IntuitionI**

nstead of one large attention mechanism, we split our Queries, Keys, and Values into $h$ smaller chunks.

- Each "Head" learns something different (e.g., one head learns grammar, another learns gender, another learns spatial relationships).
- We then concatenate all these heads back together and pass them through a final Linear layer.

2. **Implementation with Keras Layers**

You don't have to write the math from scratch every time; Keras provides a highly optimized MultiHeadAttention layer.

```python
import tensorflow as tf
from tensorflow.keras.layers import MultiHeadAttention, Input

# Hyperparameters
num_heads = 8      # Number of parallel attention heads
key_dim = 64       # Dimension of each head
batch_size = 32
seq_len = 50
embed_dim = 512    # Total embedding dimension

# 1. Define inputs
# Shape: (Batch, Sequence_Length, Embedding_Dim)
target = Input(shape=(seq_len, embed_dim))
source = Input(shape=(seq_len, embed_dim))

# 2. Initialize Multi-Head Attention layer
mha_layer = MultiHeadAttention(num_heads=num_heads, key_dim=key_dim)

# 3. Apply Attention
# In Self-Attention, query, key, and value are all the SAME input
attn_output, attn_weights = mha_layer(
    query=target, 
    value=source, 
    key=source, 
    return_attention_scores=True
)

print(f"Attention Output Shape: {attn_output.shape}") 
# Expected: (32, 50, 512)
```

3. **Why Multi-Head?**
- Parallelization: Unlike RNNs, which process word-by-word, Multi-Head Attention processes all words at once. This is why Transformers are so much faster to train.

- Rich Representation: It allows the model to "attend" to information from different representation subspaces at different positions.
