#BERT — Bidirectional Encoder Representations from Transformers

Yesterday, we built the Transformer Block. Today, we look at how Google used that block to create BERT, the model that effectively "killed" RNNs for text understanding.

Unlike previous models that read text left-to-right (like GPT) or right-to-left, BERT reads the entire sequence at once in both directions. This allows it to understand the full context of a word based on its surroundings.

1. **The Two Secrets of BERT Pre-training**

BERT wasn't just a better architecture; it introduced a new way of training models:

1. **Masked Language Modeling (MLM):** BERT hides (masks) 15% of the words in a sentence and tries to guess them.

- Example: "The [MASK] sat on the mat" → BERT guesses "cat".

2. **Next Sentence Prediction (NSP):** BERT looks at two sentences and decides if the second one logically follows the first.

2. **Implementing BERT with Hugging Face transformers**

Instead of building BERT from scratch (which costs millions of dollars in compute), we use Transfer Learning. We download the pre-trained weights and fine-tune them.

```python
# !pip install transformers datasets
from transformers import BertTokenizer, TFBertModel
import tensorflow as tf

# 1. Load the pre-trained Tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# 2. Tokenize a sample sentence
text = "Machine Learning is changing the world."
inputs = tokenizer(text, return_tensors="tf", padding=True, truncation=True, max_length=128)

# 3. Load the pre-trained BERT Model
bert_model = TFBertModel.from_pretrained('bert-base-uncased')

# 4. Get the embeddings
outputs = bert_model(inputs)

# Last Hidden State: Embeddings for every word in the sentence
# Pooled Output: A single vector representing the entire sentence (useful for classification)
print("Shape of word embeddings:", outputs.last_hidden_state.shape)
```

3. **Fine-tuning Strategy**
When you use BERT for your own projects (like your Online Fake News Detection), you:

- Keep BERT's weights mostly frozen.

- Add one Dense layer on top of the pooled_output.

- Train only the final layer on your specific data.
