#GPT (Generative Pre-trained Transformer) — The Art of Prediction

While BERT (Day 97) is an Encoder-only model designed to understand text, GPT is a Decoder-only model designed to generate text.

The core difference is the Direction of Sight:

- BERT looks at the whole sentence at once (Bidirectional).

- GPT is "Auto-regressive." It only looks at the words to its left. It tries to predict the next word, one by one.

1. **Causal Masking — The GPT Secret**

In a Transformer block, GPT uses Masked Self-Attention. This ensures the model doesn't "cheat" by looking at the next word during training. If the sentence is "The cat sat," when predicting "sat," the model is physically blocked from seeing any words after "cat."

2. **Generating Text with GPT-2**

We’ll use a smaller version (GPT-2) from Hugging Face to see how it "dreams" up text.

```python
from transformers import TFGPT2LMHeadModel, GPT2Tokenizer

# 1. Load pre-trained GPT-2
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = TFGPT2LMHeadModel.from_pretrained("gpt2")

# 2. Encode input text
input_text = "The future of Artificial Intelligence is"
input_ids = tokenizer.encode(input_text, return_tensors='tf')

# 3. Generate text
# max_length: how many tokens to generate
# do_sample: introduces variety so it's not repetitive
output = model.generate(input_ids, max_length=50, num_beams=5, no_repeat_ngram_size=2, early_stopping=True)

# 4. Decode and Print
print(tokenizer.decode(output[0], skip_special_tokens=True))
```

### **Model Comparison: BERT vs. GPT**

| Feature | BERT | GPT |
| :--- | :--- | :--- |
| **Architecture** | Transformer Encoder | Transformer Decoder |
| **Context** | Bidirectional (Left & Right) | Unidirectional (Left only) |
| **Best For** | Classification, QA, NER | Creative Writing, Chatbots |
| **Training Task** | Guess missing words (Masked) | Predict next word (Causal) |
| **Paradigm** | Auto-Encoding | Auto-Regressive |
