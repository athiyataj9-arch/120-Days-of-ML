#The Hugging Face Pipeline — NLP "Easy Mode"

After days of looking at $Q, K, V$ and complex layers, today we learn how to use these SOTA models in production with just 3 lines of code. The pipeline() abstraction handles tokenization, model loading, and post-processing automatically.

```python
from transformers import pipeline

# 1. Sentiment Analysis
classifier = pipeline("sentiment-analysis")
print(classifier("This 120-day challenge is absolutely life-changing!"))

# 2. Named Entity Recognition (NER)
ner = pipeline("ner", grouped_entities=True)
print(ner("Athiya Taj is a Data Scientist working in Bangalore, India."))

# 3. Summarization
summarizer = pipeline("summarization")
long_text = "Your long project description about the Machine Health Predictor..."
print(summarizer(long_text, max_length=50, min_length=10))
```
