**The Fine-Tuning Strategy** 

Instead of training millions of parameters from scratch (which is expensive and slow), we use Transfer Learning:

- **Load:** Take a pre-trained bert-base-uncased model.
- **Attach:** Add a classification head (a simple Dense layer) on top of the "Pooled Output.
- **Train:** Fine-tune with a very low learning rate (e.g., $2 \times 10^{-5}$) so we don't destroy the knowledge BERT already has.

1. **mplementation**

```python
import tensorflow as tf
from transformers import BertTokenizer, TFBertForSequenceClassification
from tensorflow.keras.optimizers import Adam

# 1. Load pre-trained assets
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
# Use 2 labels (e.g., Fake vs. Real News)
model = TFBertForSequenceClassification.from_pretrained('bert-base-uncased', num_labels=2)

# 2. Sample data preparation
texts = ["The new vaccine shows 95% efficacy.", "Aliens have taken over the city hall!"]
labels = [0, 1] # 0: Reliable, 1: Fake

# 3. Tokenization (Attention Mask is automatically generated)
inputs = tokenizer(texts, padding=True, truncation=True, return_tensors='tf')

# 4. Compile with a small learning rate
optimizer = Adam(learning_rate=2e-5)
loss = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)
model.compile(optimizer=optimizer, loss=loss, metrics=['accuracy'])

# 5. Training
model.fit(dict(inputs), tf.constant(labels), epochs=3)
```
