#Callbacks & Early Stopping

**Logic:** Training a model for 1,000 epochs is dangerous. If the model starts "memorizing" the training data, its performance on real-world data (Validation Loss) will actually start getting worse. Early Stopping acts as a referee that blows the whistle the moment the model stops improving.

We will use the EarlyStopping callback. This is the "Safety Net" for every professional Deep Learning pipeline.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.callbacks import EarlyStopping
import numpy as np
from sklearn.model_selection import train_test_split

# 1. Create dummy dataset
X = np.random.rand(500, 20)   # 500 samples, 20 features
y = np.random.randint(0, 2, 500)  # Binary labels

# Split into train and validation
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.2)

# 2. Create Model (fixed input layer)
model = Sequential([
    Input(shape=(20,)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# 3. Early Stopping
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True,
    verbose=1
)

# 4. Train Model
print("Day 65: Training initialized with Early Stopping.")

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=100,
    callbacks=[early_stop]
)
```
