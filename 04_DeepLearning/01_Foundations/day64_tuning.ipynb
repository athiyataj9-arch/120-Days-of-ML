#Hyperparameter Tuning (The Search for the Sweet Spot)

**Logic:** If building a model is like cooking, hyperparameters are the temperature and the timer. If the heat (Learning Rate) is too high, the food burns (the model overshoots). If it's too low, it takes forever to cook (the model never learns).

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

# 1. Define a function to create a model with a specific Learning Rate
def build_model(learning_rate):
    model = Sequential([
        Dense(64, activation='relu', input_dim=20),
        Dense(1, activation='sigmoid')
    ])
    # The Learning Rate is a critical Hyperparameter
    optimizer = Adam(learning_rate=learning_rate)
    model.compile(optimizer=optimizer, loss='binary_crossentropy', metrics=['accuracy'])
    return model

# 2. Experiment A: Fast Learning (High LR) - Risk of instability
model_fast = build_model(learning_rate=0.1)
print("Experiment A: Learning Rate 0.1 (Fast)")

# 3. Experiment B: Steady Learning (Standard LR) - More stable
model_steady = build_model(learning_rate=0.001)
print("Experiment B: Learning Rate 0.001 (Steady)")

# 4. Batch Size: Another Hyperparameter
# Smaller Batch = More updates, but slower training
# Larger Batch = Faster training, but uses more memory
batch_size_options = [16, 32, 64]

print(f"Day 64: Ready to tune with Batch Sizes: {batch_size_options}")
```
