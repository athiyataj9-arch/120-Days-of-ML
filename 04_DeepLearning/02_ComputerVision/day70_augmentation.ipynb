#Data Augmentation (Faking it 'til you make it)

**Logic:** CNNs are smart, but they can be "fooled" if an object is tilted or zoomed differently than what they saw in training. Data Augmentation creates "synthetic" variety. We take one picture of a cat and rotate it, flip it, and zoom it. Now the model has 10 pictures of a cat!

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# This creates a "Preprocessing" layer that stays with the model
data_augmentation = tf.keras.Sequential([
  layers.RandomFlip("horizontal_and_vertical"),
  layers.RandomRotation(0.2),
  layers.RandomZoom(0.1),
])

# Example of how to add it to your MNIST model
model = models.Sequential([
    data_augmentation, # The model "hallucinates" new data every epoch
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(10, activation='softmax')
])

print("Day 70: Data Augmentation layer is active.")
```
