#Pooling & Padding (Focus & Boundaries)

**Logic:**
- **Max Pooling:** This is like "summarizing" the image. It looks at a 2x2 area and only keeps the brightest pixel. This reduces the data size and makes the model faster.

- **Padding:** When the filter slides, it hits the edges less often than the center. Padding adds a "border" of zeros so the model doesn't "forget" the information at the corners.

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D

model = Sequential([
    # Convolution finds the pattern
    Conv2D(32, (3, 3), padding='same', activation='relu', input_shape=(28, 28, 1)),
    
    # Max Pooling summarizes and shrinks the image by half
    MaxPooling2D(pool_size=(2, 2))
])

model.summary()
```
