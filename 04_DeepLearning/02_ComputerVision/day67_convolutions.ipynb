#Convolutional Layers (Finding Patterns)

**Logic:** A normal neural network treats every pixel as independent. A Convolutional Neural Network (CNN) understands that pixels close to each other form shapes. It uses a Kernel (Filter)—a small window that slides across the image to find edges, curves, and textures.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D

# Imagine an input image of 28x28 pixels (Grayscale = 1 channel)
model = Sequential([
    # 32 Filters: The model will look for 32 different types of patterns
    # (3, 3): The size of the sliding window
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1))
])

model.summary()
```
