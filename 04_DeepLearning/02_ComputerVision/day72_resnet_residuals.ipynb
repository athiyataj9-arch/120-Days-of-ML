#ResNet & Skip Connections

**Logic:** Imagine you are giving directions to someone. VGG16 is like a long chain of people whispering the directions; by the end, the message is garbled. ResNet adds a "shortcut" where the original message is passed directly to later layers, ensuring the core information is never lost.

```python
import tensorflow as tf
from tensorflow.keras.applications import ResNet50
from tensorflow.keras import layers, models

# 1. Load the ResNet50 "Expert" 
# ResNet is much more efficient than VGG16
base_model_resnet = ResNet50(weights='imagenet', include_top=False, input_shape=(224, 224, 3))

# 2. Freeze the base
base_model_resnet.trainable = False

# 3. Build the Architecture
model = models.Sequential([
    base_model_resnet,
    layers.GlobalAveragePooling2D(), # More modern than Flatten() for deep models
    layers.Dense(512, activation='relu'),
    layers.Dropout(0.4),
    layers.Dense(1, activation='sigmoid') 
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.summary()
```
