#YOLO Architecture (Real-Time Detection)

**Logic:** Older detection models (like R-CNN) were slow because they looked at thousands of tiny patches of an image one by one. YOLO treats the whole image as a single regression problem. It divides the image into a grid (e.g., $13 \times 13$) and, in a single "glance," predicts bounding boxes and class probabilities for every cell.

#Loading a Pre-trained YOLOv8 Model

Instead of building YOLO from scratch (which takes months), we use the ultralytics library to load a model that already knows how to detect 80 different objects (cars, people, dogs, etc.).

```python
import sys
!"{sys.executable}" -m pip install ultralytics opencv-python --upgrade
```

```python
from ultralytics import YOLO

# 1. Load the Segmentation version of the model (Note the '-seg' suffix)
model = YOLO('yolov8n-seg.pt') 

# 2. Predict on an image
# This will detect objects AND create a pixel-perfect mask for them
results = model.predict('https://ultralytics.com/images/bus.jpg', save=True)

# 3. Accessing the Masks
for result in results:
    masks = result.masks  # This contains the polygon coordinates of the shapes
    if masks:
        print(f"Day 76: Successfully generated masks for {len(masks)} objects.")
```
