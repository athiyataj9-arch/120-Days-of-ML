#Object Detection Foundations (IoU & Bounding Boxes)

**Logic:** In classification, we output one label. In detection, the model must output four numbers for every object: [x_center, y_center, width, height]. To know if our box is "good," we use IoU (Intersection over Union). If our predicted box overlaps perfectly with the real box, IoU is 1.0.

**Calculating IoU**

Since Object Detection is mathematically heavy, today we focus on the metric that governs all detection models (YOLO, SSD, R-CNN).

```python
import numpy as np

def calculate_iou(boxA, boxB):
    # box = [x1, y1, x2, y2]
    # 1. Determine the coordinates of the intersection rectangle
    xA = max(boxA[0], boxB[0])
    yA = max(boxA[1], boxB[1])
    xB = min(boxA[2], boxB[2])
    yB = min(boxA[3], boxB[3])

    # 2. Compute the area of intersection
    interArea = max(0, xB - xA + 1) * max(0, yB - yA + 1)

    # 3. Compute the area of both bounding boxes
    boxAArea = (boxA[2] - boxA[0] + 1) * (boxA[3] - boxA[1] + 1)
    boxBArea = (boxB[2] - boxB[0] + 1) * (boxB[3] - boxB[1] + 1)

    # 4. Compute IoU (Intersection / (Union))
    iou = interArea / float(boxAArea + boxBArea - interArea)
    return iou

# Example: Box 1 (Ground Truth) and Box 2 (Prediction)
gt_box = [50, 50, 150, 150]
pred_box = [60, 60, 160, 160]

print(f"Day 74: IoU Score is {calculate_iou(gt_box, pred_box):.4f}")
```
