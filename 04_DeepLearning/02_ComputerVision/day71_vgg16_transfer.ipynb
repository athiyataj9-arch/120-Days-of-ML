#Transfer Learning Theory (VGG16)

**Logic:** Most images share basic features (edges, circles, textures). Models like VGG16 (built by Oxford) already know how to "see" these perfectly. We "freeze" their layers so they don't change, and we only train a new "head" (the final layers) for our specific classes.

```python
import tensorflow as tf
from tensorflow.keras.applications import VGG16
from tensorflow.keras import layers, models

# 1. Load the "Expert" (VGG16) without the top classification layer
# include_top=False means we drop the 1000-class head and keep the "eyes"
base_model = VGG16(weights='imagenet', include_top=False, input_shape=(150, 150, 3))

# 2. FREEZE the expert's brain so we don't ruin what it already knows
base_model.trainable = False

# 3. Build your custom "Head" on top
model = models.Sequential([
    base_model,
    layers.Flatten(),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(1, activation='sigmoid') # For Binary Classification (e.g., Cat vs Dog)
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.summary()
```
